# Procesamiento Naive vs Streaming

**Monitorización de la velocidad de la parrilla en una carrera de Fórmula 1**

Durante una carrera de Fórmula 1, cada coche mide su velocidad instantánea mediante sensores internos a una frecuencia de muestreo extremádamente alta.
Se plantean diferentes casos de uso hipotéticos para cuando estos datos llegan continuamente para todos los coches de la parilla.

## Importar librerías

```python
import random
import sys
import time
```

## Funciones

```python
def get_f1_speed():
    # Velocidades de Fórmula 1 en km/h (100-350 km/h)
    return random.uniform(100, 350)  
```

## Máxima velocidad de la parrilla

### Procesamiento naive

Ya se procesaron 10M datos históricos almacenados en una lista.

Cuando llegan nuevos datos, debe recalcular el máximo sobre TODA la lista (10M elementos), lo cual es costoso en tiempo y memoria.

```python
print("=== PROCESAMIENTO NAIVE ===")
print("Iniciando carrera...\n")

# Velocidades de Fórmula 1 en km/h
speed = [get_f1_speed() for _ in range(10000000)]

print("Procesando nuevos datos que llegan continuamente...")
print(f"{'Iteración':<12} {'Elementos':<12} {'Memoria (MB)':<15} {'Tiempo (ms)':<15} {'Máximo':<10}")
print("-" * 70)

for i in range(10):
    start = time.time()

    speed.append(get_f1_speed())
    max_speed = max(speed)
    
    elapsed = (time.time() - start) * 1000
    memory_mb = sys.getsizeof(speed) / 1024 / 1024
    print(f"{i+1:<12} {len(speed):<12,} {memory_mb:<15.2f} {elapsed:<15.2f} {max_speed:.2f}")

print(f"\nMemoria final: {sys.getsizeof(speed)/1024/1024:.2f} MB")
print(f"Elementos totales: {len(speed):,}")
```

### Procesamiento en streaming

Solo se mantiene el agregado de los anteriores datos históricos (máximo actual).

Cuando llegan nuevos datos, solo actualiza el agregado en tiempo constante, sin necesidad de recorrer todo el histórico.

```python
print("=== STREAMING PROCESSING ===")
print("Inicializando carrera ...\n")

max_speed = 0

# Velocidades de Fórmula 1 en km/h
for _ in range(10000000):
    max_speed = max(get_f1_speed(), max_speed)

print("Procesando nuevos datos que llegan continuamente...")
print(f"{'Iteración':<12} {'Procesados':<12} {'Memoria (bytes)':<18} {'Tiempo (ms)':<15} {'Máximo':<10}")
print("-" * 75)

for i in range(10):
    start = time.time()
    
    max_speed = max(get_f1_speed(), max_speed)

    elapsed = (time.time() - start) * 1000
    memory_bytes = sys.getsizeof(max_speed)
    print(f"{i+1:<12} {1:<12,} {memory_bytes:<18} {elapsed:<15.2f} {max_speed:.2f}")

print(f"\nMemoria final: {sys.getsizeof(max_speed)} bytes")
print(f"Elementos totales: {1:,}")
```

## Velocidad media de la parrilla

### Procesamiento naive

Ya se procesaron 10M datos históricos almacenados en una lista.

Cuando llegan nuevos datos, debe recalcular la media sobre TODA la lista (10M+ elementos) usando `sum(speed) / len(speed)`, lo cual es costoso en tiempo y memoria.

```python
print("=== PROCESAMIENTO NAIVE ===")
print("Iniciando carrera...\n")

# Velocidades de Fórmula 1 en km/h
speed = [get_f1_speed() for _ in range(10000000)]

print("Procesando nuevos datos que llegan continuamente...")
print(f"{'Iteración':<12} {'Elementos':<12} {'Memoria (MB)':<15} {'Tiempo (ms)':<15} {'Media':<10}")
print("-" * 70)

for i in range(10):
    start = time.time()

    speed.append(get_f1_speed())
    avg_speed = sum(speed) / len(speed)
    
    elapsed = (time.time() - start) * 1000
    memory_mb = sys.getsizeof(speed) / 1024 / 1024
    print(f"{i+1:<12} {len(speed):<12,} {memory_mb:<15.2f} {elapsed:<15.2f} {avg_speed:.2f}")

print(f"\nMemoria final: {sys.getsizeof(speed)/1024/1024:.2f} MB")
print(f"Elementos totales: {len(speed):,}")
```

### Procesamiento en streaming

Ya se procesaron 10M datos históricos, pero solo se mantiene el agregado (suma total + contador).

Cuando llegan nuevos datos, solo actualiza la suma y el contador en tiempo constante, calculando la media como `total/count` sin necesidad de recorrer todo el histórico.

```python
print("=== STREAMING PROCESSING ===")
print("Inicializando carrera ...\n")

total_speed = 0
count = 0

# Velocidades de Fórmula 1 en km/h
for _ in range(10000000):
    total_speed += get_f1_speed()
    count += 1

avg_speed = total_speed / count

print("Procesando nuevos datos que llegan continuamente...")
print(f"{'Iteración':<12} {'Procesados':<12} {'Memoria (bytes)':<18} {'Tiempo (ms)':<15} {'Media':<10}")
print("-" * 75)

for i in range(10):
    start = time.time()
    
    total_speed += get_f1_speed()
    count += 1
    avg_speed = total_speed / count

    elapsed = (time.time() - start) * 1000
    memory_bytes = sys.getsizeof(total_speed) + sys.getsizeof(count)
    print(f"{i+1:<12} {1:<12,} {memory_bytes:<18} {elapsed:<15.2f} {avg_speed:.2f}")

print(f"\nMemoria final: {sys.getsizeof(total_speed) + sys.getsizeof(count)} bytes")
print(f"Elementos totales: {1:,}")
```
